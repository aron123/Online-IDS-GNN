```python
path_X_test = '/mnt/d/TON_IoT_Dataset/complete/X_test.csv'
path_X_train = '/mnt/d/TON_IoT_Dataset/complete/X_train.csv'
path_orig_dataset = '/mnt/d/TON_IoT_Dataset/combined.csv'
path_output = '/mnt/d/TON_IoT_Dataset/complete/_X_test.csv'
```

```python
orig_df = pd.read_csv(path_orig_dataset, usecols=['label', 'ts'])
test_df = pd.read_csv(path_X_test)
```

```python
split_millisecs = 20
start_time = orig_df['ts'].min()
orig_df['time_group'] = (orig_df['ts'] - start_time) // split_millisecs
orig_df = orig_df.reset_index(drop=True).sort_values('time_group')
```

```python
test_time_groups = test_df['time_group'].unique()
orig_df_filtered = orig_df[orig_df['time_group'].isin(test_time_groups)]
orig_df_filtered
```
